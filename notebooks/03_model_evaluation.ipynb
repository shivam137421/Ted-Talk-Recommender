### Objective
The objective of this milestone is to evaluate the performance of the content-based TED Talks recommendation system using suitable evaluation metrics and recommendation analysis.

### Evaluation Approach
The system will be evaluated using its TF-IDF representations, cosine similarity scores, and recommendation results.

### Evaluation Metrics
- Precision@K
- Recall@K
- Mean Average Precision (MAP)
- Normalized Discounted Cumulative Gain (NDCG)

*Note: Ranking metrics require relevant-item judgments or suitable ground-truth data. These will be established before interpreting evaluation results.

```python
import pandas as pd
import numpy as np
import joblib
from pathlib import Path

# Set model directory
MODEL_DIR = Path("../models")

# Load saved training and testing datasets
train_df = joblib.load(MODEL_DIR / "train_dataset.pkl")
test_df = joblib.load(MODEL_DIR / "test_dataset.pkl")

# Load the test-to-train similarity matrix
test_train_similarity = joblib.load(
    MODEL_DIR / "test_train_similarity.pkl"
)

# Display dataset and similarity matrix information
print("Training Dataset Shape:", train_df.shape)
print("Testing Dataset Shape:", test_df.shape)
print("Similarity Matrix Shape:", test_train_similarity.shape)
```
